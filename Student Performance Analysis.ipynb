```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "Student":["Ram","Sita","Avinash","Ravi","Madhu","Anu","Sam"],
    "Roll no":[101,102,103,104,105,106,107],
    "Sub 1":[85,92,24,45,18,70,88],
    "Sub 2":[90,28,55,54,75,60,80],
    "Sub 3":[80,35,60,44,64,80,85]
}
dataset=pd.DataFrame(data)
print("Data set:")
print(dataset)


#caluculating total for each student
dataset["Total"]=dataset["Sub 1"]+dataset["Sub 2"]+dataset["Sub 3"]


#caluculating average for each student
dataset["Average"]=dataset["Total"]/3
print("\n")
print( dataset)


#checking whether student is passed or failed
dataset["Result"] = (
    (dataset["Sub 1"] >= 35) &
    (dataset["Sub 2"] >= 35) &
    (dataset["Sub 3"] >= 35)
).map({True: "Pass", False: "Fail"})
print("\n")
print(dataset)

#weak performance of students

subjects = ["Sub 1", "Sub 2", "Sub 3"]

averages = dataset[subjects].mean()

weak_subject = averages.idxmin()
print("\n")
print("Weak Subject:", weak_subject)
print("Average Marks:", averages[weak_subject])

# average representation
plt.bar(dataset["Student"], dataset["Average"])

plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.title("Student Performance Analysis")

plt.show()

#pie representation of result
print("\n")

for subject in subjects:
    high = (dataset[subject] >= 75).sum()
    low = (dataset[subject] < 75).sum()

    plt.figure(figsize=(5, 5))

    plt.pie(
        [high, low],
        labels=["High Score", "Low Score"],
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title(subject + " - Student Performance")
    plt.show()
```
